## 1. Import Libraries and Load Dataset

In this section, we import the required Python libraries and load the raw customer personality dataset into a Pandas DataFrame.

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv(r"C:\Users\mudas\OneDrive\Documents\Data_Analyst_Internship_Task_1\Dataset\marketing_campaign.csv", sep="\t")

# Display basic information
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

## 2. Initial Dataset Inspection

Inspect the dataset structure, column names, data types, and statistical summary to understand the raw data.

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

## 3. Handling Missing Values and Duplicates

Identify missing values, replace missing income values with the median, and check for duplicate records.

```python
# Check missing values
print("Missing values before cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# Fill missing Income values with median
median_income = df['Income'].median()
df['Income'] = df['Income'].fillna(median_income)

# Check and remove duplicates
print("\nDuplicate rows:", df.duplicated().sum())
df.drop_duplicates(inplace=True)

# Verify
print("\nMissing values after cleaning:", df.isnull().sum().sum())
print("Dataset shape:", df.shape)
```

## 4. Standardizing Text and Column Names

Remove unnecessary spaces, standardize text formatting, and convert column names into lowercase.

```python
# Standardize column names
df.columns = df.columns.str.lower().str.strip()

# Standardize text columns
df['education'] = df['education'].str.strip().str.title()
df['marital_status'] = df['marital_status'].str.strip().str.title()

# Restore original category formats
df['education'] = df['education'].replace({
    'Phd': 'PhD',
    '2N Cycle': '2n Cycle'
})

df['marital_status'] = df['marital_status'].replace({
    'Yolo': 'YOLO'
})

# Display standardized values
print("Education:", df['education'].unique())
print("Marital Status:", df['marital_status'].unique())
print("\nColumn names:", df.columns.tolist())
```

## 5. Correcting Data Types

Convert the customer registration date into the proper datetime format and verify column data types.

```python
# Convert date column
df['dt_customer'] = pd.to_datetime(
    df['dt_customer'],
    format='%d-%m-%Y'
)

# Display data types
print(df.dtypes)
```

## 6. Outlier Detection and Treatment

Use the Interquartile Range (IQR) method to identify potential outliers in the Income column. Investigate unusually high values and replace the suspicious value 666666 with the median.

```python
# Calculate IQR
Q1 = df['income'].quantile(0.25)
Q3 = df['income'].quantile(0.75)
IQR = Q3 - Q1

# Calculate outlier limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[(df['income'] < lower_limit) | (df['income'] > upper_limit)]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Potential Outliers:", len(outliers))
print(outliers[['id', 'income']])

# Replace suspicious income value
median_income = df['income'].median()
df.loc[df['income'] == 666666, 'income'] = median_income

print("\nMaximum Income After Treatment:", df['income'].max())
```

## 7. Final Data Validation

Verify that the dataset is clean by checking missing values, duplicate records, dataset dimensions, and data types.

```python
print("Final Dataset Shape:", df.shape)
print("Total Missing Values:", df.isnull().sum().sum())
print("Total Duplicate Rows:", df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)

print("\nCleaned Dataset Preview:")
display(df.head())
```

## 8. Exporting the Cleaned Dataset

Save the cleaned dataset as a CSV file for submission.

```python
df.to_csv("cleaned_marketing_campaign.csv", index=False)

print("Cleaned dataset exported successfully!")
```
